## MCP Servers: Local vs Remote Availability

**Legend**
- 🌐 **Remote**: the vendor hosts the server; you just add the URL to your MCP client (usually signing in with OAuth).
- 💻 **Local**: you run the server yourself, either through a stdio command (`npx`, `uvx`, `docker`) or as a local HTTP server.
- ❌: no official option of that type.

| Server | 🌐 Remote (vendor-hosted) | 💻 Local | Docs |

| **Notion** | ✅ `https://mcp.notion.com/mcp` (OAuth) | ✅ `npx -y @notionhq/notion-mcp-server` (integration token). | **PayPal** | ✅ `https://mcp.paypal.com/mcp`; sandbox at `https://mcp.sandbox.paypal.com/mcp` | ✅ `npx -y @paypal/mcp --tools=all` (access token + `PAYPAL_ENVIRONMENT`) | [mcp.paypal.com](https://mcp.paypal.com) · [Agent Toolkit](https://github.com/paypal/agent-toolkit) |

# <span style='background:lightgreen'>4. **Claude Desktop**</span>

## a. Overview

<h2 align="center"><div class="alert alert-success" style="margin: 20px"><b>Claude Desktop</b> is a native desktop application by Anthropic for <b>macOS</b> and <b>Windows</b> that unifies Claude Chat, Claude Code, and Claude Cowork in a single interface — giving you AI-powered assistance both in conversation and as an autonomous agent that can act on your computer.</div></h2>

**Key highlights:**
- Supports Anthropic's Claude models (Opus, Sonnet, Haiku) as well as locally or cloud-hosted third-party models via Ollama (e.g., Qwen, GLM, Kimi) and providers such as LM Studio and llama.cpp.
- Connects to local and remote tools, files, and apps via **Desktop Extensions** (local) and **Web Connectors** (cloud-based).
- Access tiers: **Claude Chat** (Free), **Claude Code** (Pro/Max/Team/Enterprise), **Claude Cowork** (Pro/Max/Team/Enterprise).

## b. Installation and First Launch

**System Requirements:**
- macOS 11 (Big Sur) or higher
- Windows 10 or higher

**Steps to install:**
1. Visit [https://claude.com/download](https://claude.com/download) and download the installer for your OS (macOS or Windows).
2. Open the downloaded file and complete the installation.
3. Launch Claude from your **Applications** folder (macOS) or **Start Menu** (Windows).
4. Sign in using your email or Google account to sync your chats and settings.

📖 Reference: [Install Claude Desktop](https://support.claude.com/en/articles/10065433-install-claude-desktop)


## c. Interface Walk-Through: The Left Pane

- Once installed, simply launch Claude Desktop; it runs in its **standard Anthropic profile** (using native Claude models such as Opus, Sonnet, and Haiku). If you have previously pointed it at Ollama (see section 4g), restore the default profile with the following command:
```bash
$ ollama launch claude-desktop --restore
```

- The left pane is your primary navigation hub. Explore each of the following:

### i. New Chat
- Click **New Chat** to begin a fresh conversation thread (independent of all prior threads by default).
- **Practice:** Start a new chat → rename it → pin it → share it using the **Share** button (top-right, generates a shareable link) → add it to a Project using **"Add to Project"**.
- **Charts and Widgets:** Charts and widgets are useful when you need to see data, ideas, or processes clearly. You don’t need an external tool. Just ask Claude to turn the information into a visualization. To enable this, go to **Settings → Capabilities → Visuals** and turn on **Inline visualizations**.

### ii. [Artifacts](https://support.claude.com/en/articles/9487310-what-are-artifacts-and-how-do-i-use-them)
- Artifacts let Claude generate shareable apps, tools, visualizations, code, and documents in a dedicated side panel — separate from the main conversation window.
- A live artifact is an artifact that connects to external data sources and refreshes with real-time data every time it's opened. A live artifact is useful when you need a dashboard, tracker, or workspace you can return to. Claude builds it once, and then you keep reviewing, updating, and reusing it inside Claude Cowork. (Available in Claude Cowork)
- Available on Free, Pro, Max, Team, and Enterprise plans.
- **Practice:** Go to **Settings → Capabilities → Enable Artifacts and AI-powered artifacts**. Then ask Claude to *"create an interactive bar chart of Pakistan's population by province"* and observe the Artifact panel open alongside the chat.


### iii. [Projects](https://support.claude.com/en/articles/9517075-what-are-projects)
- Projects are self-contained workspaces with their own chat histories, uploaded knowledge bases, and custom instructions — ideal for recurring or long-running work.
- **Practice:** Create a new Project called *"AI Research"* → upload a PDF paper → set a custom system prompt → start a focused chat within the project and observe how Claude retains context across sessions.

### iv. [Claude Code](https://support.claude.com/en/articles/11145838-use-claude-code-with-your-pro-or-max-plan)
- A command-line agentic coding tool integrated into Claude Desktop (Pro/Max/Team/Enterprise). It allows you to delegate complex coding, debugging, and repository tasks directly from your terminal while maintaining full transparency and control.
- **Note:** On the Free tier, this option is locked. On Pro and above, Claude Code and Claude Chat share a unified subscription.
- **Practice (Pro/Max):** Click the **Code** tab in the left pane → select a project folder → ask Claude to *"read my project folder and explain its structure"*.

### v. [Claude Cowork](https://support.claude.com/en/articles/13345190-get-started-with-claude-cowork)
- Cowork brings Claude Code's agentic architecture to a **visual, no-terminal interface** for knowledge work (documents, research, file management, spreadsheets, and more). Claude analyzes your request, builds a plan, breaks it into subtasks, and executes them — with full transparency at each step.
- Requires: Claude Desktop app + paid plan (Pro, Max, Team, or Enterprise) + active internet connection + the app must remain **open** while Claude works.
- **Practice:** Click the **Cowork** tab → describe a multi-step task such as *"Summarize all PDF files in my Downloads folder into a single Word document"* → review Claude's plan → let it execute → inspect the output.

### vi. Pinned & Recent Chats
- Previously active or pinned chats appear in the **Pinned** and **Recent** sections for quick access.

### vii. Avatar (Settings)
- Click your avatar at the bottom of the left pane to access **Settings**, where you can see:
    - **General:** Enter your personal preferences (name, what Claude should call you), enable/disable response completion notifications, and choose appearance and voice settings.
    - **Account:** View your account details (name, email), manage sessions, and log out or delete your account.
    - **Privacy:** Control whether your chats may be used to improve models, manage location metadata, and export your data.
    - **Billing:** See your current plan and usage, upgrade or cancel, and view invoices.
    - **Capabilities:** Toggle features such as Artifacts, Memory, Web Search, Inline visualizations, and code execution/file creation.
    - **Connectors:** Add, manage, and remove remote web connectors (directory and custom connectors).
    - **Extensions:** Browse, install, and configure local Desktop Extensions (`.mcpb`).
    - **Developer:** Edit `claude_desktop_config.json` to register local MCP servers and view their logs.
    - **Claude Code:** Configure Claude Code inside the desktop app (permissions, default folders, models).
- **Practice:** Explore **Settings → Capabilities** (toggle Artifacts, Web Search, Memory) and **Settings → Extensions** (browse Desktop Extensions).

## g. Running Claude Desktop with Ollama

### Overview
- Claude Desktop can be integrated with Ollama Cloud, enabling the use of Ollama-hosted models — including within Claude Cowork and Claude Code inside the app. This is particularly useful for students and practitioners who wish to leverage powerful third-party models as alternatives to Anthropic's native models (Opus, Sonnet, Haiku), often at significantly lower cost or entirely free of charge.
- **Important Note on Claude Cowork and Claude Code:** These two features within Claude Desktop are available to all users once the Ollama integration is configured. You do **not** necessarily need to upgrade to a paid Anthropic subscription plan to use them via Ollama — the integration routes inference through Ollama's infrastructure rather than Anthropic's API.
- If you are on a budget or wish to experiment with open-weight models, using Claude Desktop via Ollama Cloud is an excellent approach. Many Ollama Cloud models are available on a **free tier with generous limits**, making this setup practical for coursework and personal projects without incurring API charges.

### Prerequisites
- Before proceeding, ensure the following requirements are met:
    - **Claude Desktop** must be installed on your machine (available for **macOS** or **Windows**).
    - **Ollama** must be installed and updated to at least **v0.15** (the `ollama launch` command was introduced in this version). You can download the latest version from [https://ollama.com/download](https://ollama.com/download).
    - **An Ollama API key** is required. Obtain yours from [https://ollama.com/settings/keys](https://ollama.com/settings/keys), then set it in your terminal shell **before** launching:
```bash
# On macOS/Linux
export OLLAMA_API_KEY=your_api_key_here

# On Windows (PowerShell)
$env:OLLAMA_API_KEY="your_api_key_here"
```

### Step-by-Step Setup

#### Step 1 — Configure Claude Desktop to Use Ollama
- Run the following command in your terminal. This performs the **quick setup** that configures Claude Desktop to use Ollama Cloud as its inference backend. The following command automatically modifies Claude Desktop's configuration so that it points to Ollama Cloud instead of Anthropic's API. No manual editing of environment variables or configuration files is needed.

```bash
$ ollama launch claude-desktop
```

#### Step 2 — Explore Available Ollama Cloud Models
- After setup, Claude Desktop **automatically discovers** your available Ollama Cloud models. For example, `kimi-k2.6` will appear inside Claude Cowork, and the same models are also available to Claude Code inside Claude Desktop.
- Some of the notable Ollama Cloud models you may encounter include:

| Model | Provider | Notes |
|---|---|---|
| `kimi-k2.6` | Moonshot AI | Strong reasoning and coding |
| `qwen3.5` | Alibaba | Multilingual, instruction-tuned |
| `minimax-m2.7` | MiniMax | Agentic coding and tool use |
| `glm-5.1` | Zhipu AI | Strong agentic coding and reasoning |

> You can browse the full model library at [https://ollama.com/library](https://ollama.com/library).

### Additional Commands

#### Configure Without Launching
- If you want to apply the configuration changes **without immediately launching** Claude Desktop, use:
```bash
$ ollama launch claude-desktop --config
```

#### Restore the Default Anthropic Profile
To switch Claude Desktop back to its **standard Anthropic profile** (using native Claude models such as Opus, Sonnet, and Haiku), run:
```bash
$ ollama launch claude-desktop --restore
```
- If Claude Desktop is already running and you wish to apply the restore without manual intervention, append the `--yes` flag to automatically approve the restart:
```bash
$ ollama launch claude-desktop --restore --yes
```

### What Is Currently Supported (and What Is Not)
- Claude Desktop with Ollama currently supports the following capabilities:
    - **Ollama Cloud** as the third-party inference gateway
    - **Automatic model discovery** from Ollama Cloud
    - **Claude Cowork** with Ollama Cloud models
    - **Claude Code** inside Claude Desktop with the same cloud models
    - **Subagents** (you can instruct Claude to have subagents inherit the current model)
- The following features are **not yet supported** in this configuration:
    - Web search
    - Extensions

### Summary of Key Commands

| Purpose | Command |
|---|---|
| Set up and launch with Ollama | `ollama launch claude-desktop` |
| Configure only (no launch) | `ollama launch claude-desktop --config` |
| Restore default Anthropic profile | `ollama launch claude-desktop --restore` |
| Restore and auto-approve restart | `ollama launch claude-desktop --restore --yes` |

## a. Ways to Connect your Claude Desktop to an MCP Server
- Claude Desktop acts as an MCP client, so it can connect to local and remote MCP servers to extend Claude's capabilities with external tools and data. By default, no external tools are connected.
- **Directory Connectors (https://claude.ai/directory/connectors):** Anthropic-reviewed remote servers listed in Claude's built-in gallery. Click **Connect**, sign in with OAuth, and you are done.
- **Custom Connectors (https://registry.modelcontextprotocol.io):** Any remote MCP server that is not in the gallery. You paste its URL yourself (**Settings → Connectors → Add custom connector**).
- **Desktop Extensions (https://github.com/modelcontextprotocol/mcpb):** Local MCP servers packaged as one-click `.mcpb` bundles and installed from **Settings → Extensions**.
- **Config File (https://mcpservers.org/):** Any local MCP server registered by hand in `claude_desktop_config.json`, giving full control over the command and its arguments.

> **Free-tier students:** You have access to many pre-built directory connectors and unlimited desktop extensions (local). You may also add **one** custom (self-hosted) remote MCP connector. Your server must be publicly reachable on the internet for custom connectors to work. For fully local/private MCP servers, use Desktop Extensions or the config file instead.

# <span style='background:lightgreen'>9. Hands-on Practice: Connecting Claude Desktop with the **Git** MCP Server using the **Config File**</span>
### https://github.com/modelcontextprotocol/servers/tree/main/src/git

> **Prerequisites:** [`uv`](https://docs.astral.sh/uv/getting-started/installation/) (which provides the `uvx` command) and Git must be installed on your machine.

- Not all MCP servers have a Connector or Extension wrapper built for them. For servers without one-click support — including any custom or private local MCP server you build yourself — you register them manually in the `claude_desktop_config.json` file.
- **Important distinction:** Local MCP servers configured via this file use **your local machine's network** directly. This is a fundamentally different mechanism from remote Connectors (which route through Anthropic's cloud). Consequently, servers registered here are **not available** in Cowork or claude.ai — only in Claude Desktop and Claude Code.

**Step 1 — Locate and open the config file:**

| OS | File path |
|---|---|
| macOS | `~/Library/Application Support/Claude/claude_desktop_config.json` |
| Windows | `%APPDATA%\Claude\claude_desktop_config.json` |

The most reliable way to open it is directly from within the app:
- **macOS:** Claude menu (top system menu bar) → **Settings → Developer → Edit Config**
- **Windows:** Inside the app → **Settings → Developer → Edit Config**

This opens (or creates) the config file in your default text editor.

**Step 2 — Add your MCP server configuration:**

Obtain the MCP server's configuration snippet from its documentation (GitHub page, npm package page, etc.) and paste it into the `mcpServers` object. Below is a complete example registering the official Git MCP server (replace `/path/to/your/repo` with the path of a local Git repository):

```json
{
  "mcpServers": {
    "git": {
      "command": "uvx",
      "args": ["mcp-server-git", "--repository", "/path/to/your/repo"]
    }
  }
}
```

Save the file after pasting.

**Step 3 — Fully quit and relaunch Claude Desktop:**

Claude Desktop reads the config file **only at startup**, so changes will not take effect until you restart the application completely:

- **macOS:** Claude menu (top system menu bar) → **Quit**, or press **Cmd+Q**, then relaunch
- **Windows:** Right-click the Claude icon in the system tray → **Quit**, or use Task Manager → End Task, then relaunch

**Step 4 — Verify the integration:**

After relaunching, click the **`+`** button at the bottom of the chat input box → **Connectors**. The tools exposed by your newly registered MCP server should appear in the list. If they do, the server is connected and fully operational.

> **Practice:** Add the official Git MCP server to your config file pointing to one of your local repositories → quit and relaunch Claude Desktop → verify the tools appear via the `+` menu → ask Claude: *"Using the git tools, show me the last 5 commits and the current status of my repository."*
